```python
#checking invalid values
print("Negative Yield:", (df['Yield_Tonnes_Ha'] < 0).sum())
print("Negative Profit:", (df['Profit_INR'] < 0).sum())
print("Negative Rainfall:", (df['Rainfall_mm'] < 0).sum())
print("Negative Farm Area:", (df['Farm_Area_Hectares'] < 0).sum())
print("Negative Revenue:", (df['Revenue_INR'] < 0).sum())
print("Negative Cost:", (df['Total_Cost_INR'] < 0).sum()
     )
```

```python
#checking categorical values
for col in ['State', 'District', 'Crop', 'Season', 'Irrigation_Method']:
    print("\n", col)
    print(df[col].value_counts())
```

```python
df['Rainfall_mm'] = df['Rainfall_mm'].fillna(
    df['Rainfall_mm'].median()
)

df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(
    df['Soil_Moisture_pct'].median()
)

df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(
    df['Yield_Tonnes_Ha'].median()
)
```

```python
print(df.isnull().sum())
```

```python
df['Calculated_Production'] = (
    df['Farm_Area_Hectares'] * df['Yield_Tonnes_Ha']
)
```

```python
df['Production_Difference'] = (
    df['Production_Tonnes'] -
    df['Calculated_Production']
)
```

```python
df['Production_Difference'].describe()
```

```python
df[
    ['Production_Tonnes',
     'Calculated_Production',
     'Production_Difference']
].head(10)
```

```python
df['Calculated_Profit'] = (
    df['Revenue_INR'] -
    df['Total_Cost_INR']
)
```

```python
df['Profit_Difference'] = (
    df['Profit_INR'] -
    df['Calculated_Profit']
)
```

```python
df[
    [
        'Revenue_INR',
        'Total_Cost_INR',
        'Profit_INR',
        'Calculated_Profit',
        'Profit_Difference'
    ]
].head(10)
```

```python
print(
    "Rows where recorded profit differs from calculated profit:",
    (abs(df['Profit_Difference']) > 1).sum()
)
```

```python
#profit margin
# Profit Margin (%)
# Formula: Profit Margin = Profit / Revenue × 100

df['Profit_Margin_pct'] = np.where(
    df['Revenue_INR'] > 0,
    (df['Profit_INR'] / df['Revenue_INR']) * 100,
    np.nan
)

print(df[['Revenue_INR', 'Profit_INR', 'Profit_Margin_pct']].head())
```

```python
# Check revenue and profit values

print("Negative Revenue:", (df['Revenue_INR'] < 0).sum())
print("Zero Revenue:", (df['Revenue_INR'] == 0).sum())
print("Negative Profit:", (df['Profit_INR'] < 0).sum())

print("\nRevenue statistics:")
print(df['Revenue_INR'].describe())

print("\nProfit statistics:")
print(df['Profit_INR'].describe())
```

```python
# Check Profit = Revenue - Cost

df['Calculated_Profit'] = (
    df['Revenue_INR'] - df['Total_Cost_INR']
)

df['Profit_Difference'] = (
    df['Profit_INR'] - df['Calculated_Profit']
)

print(
    df[
        ['Revenue_INR',
         'Total_Cost_INR',
         'Profit_INR',
         'Calculated_Profit',
         'Profit_Difference']
    ].head(10)
)
```

```python
df.info()
```

```python
#revenue per hectare
df['Revenue_per_Hectare'] = (
    df['Revenue_INR'] / df['Farm_Area_Hectares']
)
```

```python
df.info()
```

```python
#cost per hectare
df['Cost_per_Hectare'] = (
    df['Total_Cost_INR'] / df['Farm_Area_Hectares']
)
```

```python
#data cleaning

df['Rainfall_mm'] = df['Rainfall_mm'].fillna(df['Rainfall_mm'].median())
df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(
    df['Soil_Moisture_pct'].median()
)
df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(
    df['Yield_Tonnes_Ha'].median()
)
```

```python
#Questions and their solutions by personal analysis
#seasonal distribution
df['Season'].value_counts()
```

```python
#visual
sns.countplot(data=df, x='Season')

plt.title('Number of Farms by Season')
plt.xlabel('Season')
plt.ylabel('Number of Farms')
plt.show()
```

```python
#average yield by season
season_yield = (
    df.groupby('Season')['Yield_Tonnes_Ha']
      .mean()
      .sort_values(ascending=False)
)

season_yield
```

```python
#visual
season_yield.plot(kind='bar')

plt.title('Average Yield by Season')
plt.xlabel('Season')
plt.ylabel('Average Yield (Tonnes/Ha)')
plt.xticks(rotation=0)
plt.show()
```

```python
#total production by season
production_season = (
    df.groupby('Season')['Production_Tonnes']
      .sum()
      .sort_values(ascending=False)
)

production_season
```

```python
#visuals
production_season.plot(kind='bar')

plt.title('Total Agricultural Production by Season')
plt.xlabel('Season')
plt.ylabel('Total Production (Tonnes)')
plt.xticks(rotation=0)
plt.show()
```

```python
#crop performance by season
crop_season_yield = pd.pivot_table(
    df,
    values='Yield_Tonnes_Ha',
    index='Crop',
    columns='Season',
    aggfunc='mean'
)

crop_season_yield
```

```python
#visual
sns.heatmap(
    crop_season_yield,
    annot=True,
    fmt='.2f'
)

plt.title('Average Crop Yield Across Seasons')
plt.xlabel('Season')
plt.ylabel('Crop')
plt.show()
```

```python
#environmental analysis
environment = df.groupby('Season')[
    [
        'Rainfall_mm',
        'Avg_Temperature_C',
        'Humidity_pct',
        'Sunlight_Hours_Day',
        'Soil_Moisture_pct'
    ]
].mean()

environment
```

```python
#visuals
environment.plot(kind='bar', figsize=(12,6))

plt.title('Environmental Conditions Across Seasons')
plt.xlabel('Season')
plt.ylabel('Average Value')
plt.xticks(rotation=0)
plt.legend()
plt.show()
```

```python
#correlation analysis
numeric_cols = [
    'Rainfall_mm',
    'Avg_Temperature_C',
    'Humidity_pct',
    'Sunlight_Hours_Day',
    'Soil_Moisture_pct',
    'Fertilizer_kg_ha',
    'Pesticide_Litre_ha',
    'Seed_Quality_Score',
    'Yield_Tonnes_Ha',
    'Production_Tonnes',
    'Profit_INR',
    'Water_Used_m3',
    'Water_Efficiency_t_per_1000m3',
    'Disease_Pest_Risk_pct'
]

corr = df[numeric_cols].corr()

plt.figure(figsize=(14,10))

sns.heatmap(
    corr,
    annot=True,
    fmt='.2f',
    cmap='coolwarm'
)

plt.title('Correlation Matrix of Agricultural Variables')
plt.show()
```

```python
#According to the given PPT questions
#Q1.How does agricultural performance vary across seasons?
season_performance = df.groupby('Season').agg(
    Avg_Yield=('Yield_Tonnes_Ha', 'mean'),
    Total_Production=('Production_Tonnes', 'sum'),
    Avg_Revenue=('Revenue_INR', 'mean'),
    Avg_Profit=('Profit_INR', 'mean'),
    Avg_Water_Efficiency=('Water_Efficiency_t_per_1000m3', 'mean')
)

season_performance
```

```python
#visuals
season_performance['Avg_Yield'].sort_values().plot(
    kind='bar'
)

plt.title('Average Agricultural Yield by Season')
plt.xlabel('Season')
plt.ylabel('Average Yield (Tonnes/Ha)')
plt.xticks(rotation=0)
plt.show()
```

```python
#Q2.what major seasonal patterns can be observed
season_pattern = df.groupby('Season').agg(
    Rainfall=('Rainfall_mm', 'mean'),
    Temperature=('Avg_Temperature_C', 'mean'),
    Humidity=('Humidity_pct', 'mean'),
    Soil_Moisture=('Soil_Moisture_pct', 'mean'),
    Yield=('Yield_Tonnes_Ha', 'mean'),
    Water_Efficiency=('Water_Efficiency_t_per_1000m3', 'mean'),
    Profit=('Profit_INR', 'mean'),
    Risk=('Disease_Pest_Risk_pct', 'mean')
)

season_pattern
```

```python
#Q3.Which characteristics change between seasons?
characteristics = df.groupby('Season')[
    [
        'Rainfall_mm',
        'Avg_Temperature_C',
        'Humidity_pct',
        'Sunlight_Hours_Day',
        'Soil_Moisture_pct',
        'Fertilizer_kg_ha',
        'Pesticide_Litre_ha',
        'Yield_Tonnes_Ha',
        'Water_Used_m3',
        'Water_Efficiency_t_per_1000m3',
        'Profit_INR',
        'Disease_Pest_Risk_pct'
    ]
].mean()

characteristics
```

```python
#Q4.What differences exist between agricultural activities in different seasons?
crop_season = pd.pivot_table(
    df,
    values='Yield_Tonnes_Ha',
    index='Crop',
    columns='Season',
    aggfunc='mean'
)

crop_season
```

```python
plt.figure(figsize=(12,7))

sns.heatmap(
    crop_season,
    annot=True,
    fmt='.2f'
)

plt.title('Crop Yield Performance Across Seasons')
plt.xlabel('Season')
plt.ylabel('Crop')

plt.show()
```

```python
#Q5.Are there noticeable variations in resource usage across seasons?
resource_usage = df.groupby('Season').agg(
    Water_Used=('Water_Used_m3', 'mean'),
    Fertilizer=('Fertilizer_kg_ha', 'mean'),
    Pesticide=('Pesticide_Litre_ha', 'mean'),
    Water_Efficiency=('Water_Efficiency_t_per_1000m3', 'mean')
)

resource_usage
```

```python
#visuals
resource_usage.plot(
    kind='bar',
    figsize=(12,6)
)

plt.title('Resource Usage Across Seasons')
plt.xlabel('Season')
plt.ylabel('Average Usage')
plt.xticks(rotation=0)
plt.show()
```

```python
#Q6.Are environmental conditions related to agricultural performance?
environment_yield = df[
    [
        'Rainfall_mm',
        'Avg_Temperature_C',
        'Humidity_pct',
        'Sunlight_Hours_Day',
        'Soil_Moisture_pct',
        'Yield_Tonnes_Ha'
    ]
].corr()

environment_yield
```

```python
#visuals
sns.heatmap(
    environment_yield,
    annot=True,
    fmt='.2f',
    cmap='coolwarm'
)

plt.title('Relationship Between Environmental Conditions and Yield')
plt.show()
```

```python
#agg
economic = df.groupby('Season').agg(
    Avg_Revenue=('Revenue_INR', 'mean'),
    Avg_Cost=('Total_Cost_INR', 'mean'),
    Avg_Profit=('Profit_INR', 'mean')
)

economic
```
